```python
import numpy as np
from kernelcpd import pelt
np.random.seed(123)
```

```python
# sample signal, 3 segments, 2 changepoints at index 29 and 59
seg_len = 30 # segment length
sd = 0.5
signal1 = np.column_stack((np.random.normal(1.0, sd, seg_len), np.random.normal(5.0, sd, seg_len)))
signal2 = np.column_stack((np.random.normal(5.0, sd, seg_len), np.random.normal(1.0, sd, seg_len)))
signal3 = np.column_stack((np.random.normal(4.0, sd, seg_len), np.random.normal(2.0, sd, seg_len)))
y = np.vstack((signal1, signal2, signal3))
```

```python
# model L2 to get 2 changepoints
print(pelt(sequence = y, pen = 2, model = 'L2'))
```

```python
# model srbf to get 2 changepoints with chunk size T = 1
print(pelt(sequence = y, pen = 4, model = 'srbf', T = 1))
```
