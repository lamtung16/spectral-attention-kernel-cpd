```python
import numpy as np
from kernelcpd import pelt, binseg
import matplotlib.pyplot as plt
np.random.seed(123)
```

Hyperparameters:

    - T: chunk size

    - sigma: kernel scale

    - laplace_option: dirichlet, tps, heat, h1

    - laplace_hyperparameter

    - kernel_type: linear, gaussian, laplace

    - centered: boolean

    - nonlinear_hyperparam

### EXAMPLE: CHANGE IN MEAN

```python
# sample signal, 3 segments, 2 changepoints at index 29 and 59
seg_len = 30 # segment length
sd = 0.5
signal1 = np.column_stack((np.random.normal(1.0, sd, seg_len), np.random.normal(5.0, sd, seg_len)))
signal2 = np.column_stack((np.random.normal(5.0, sd, seg_len), np.random.normal(1.0, sd, seg_len)))
signal3 = np.column_stack((np.random.normal(4.0, sd, seg_len), np.random.normal(2.0, sd, seg_len)))
y = np.vstack((signal1, signal2, signal3))
```

```python
plt.figure()
plt.plot(y[:, 0])
plt.plot(y[:, 1])
plt.xlabel("Time step")
plt.ylabel("y")
plt.show()
```

```python
pelt(sequence = y, 
     pen = 2, 
     model = 'L2')
```

```python
pelt(sequence = y, 
     pen = 3.5, 
     model = 'srbf', 
     T = 1, 
     sigma = 1,
     laplace_option = 'dirichlet',
     laplace_hyperparameter = 1,
     kernel_type = 'linear',
     centered = False,
     nonlinear_hyperparam = 1)
```

```python
binseg( sequence = y, 
        n_changepoints = 2, 
        model = 'srbf', 
        T = 1, 
        sigma = 1,
        laplace_option = 'dirichlet',
        laplace_hyperparameter = 1,
        kernel_type = 'linear',
        centered = False,
        nonlinear_hyperparam = 1)
```

### EXAMPLE: CHANGE IN VARIANCE

```python
# sample signal, 3 segments, 2 changepoints at index 29 and 59
seg_len = 30 # segment length
signal1 = np.column_stack((np.random.normal(1.0, 0.1, seg_len), np.random.normal(5.0, 0.1, seg_len)))
signal2 = np.column_stack((np.random.normal(1.0, 2.0, seg_len), np.random.normal(5.0, 2.0, seg_len)))
signal3 = np.column_stack((np.random.normal(1.0, 0.1, seg_len), np.random.normal(5.0, 0.1, seg_len)))
y = np.vstack((signal1, signal2, signal3))
```

```python
plt.figure()
plt.plot(y[:, 0])
plt.plot(y[:, 1])
plt.xlabel("Time step")
plt.ylabel("y")
plt.show()
```

```python
pelt(sequence = y, 
     pen = 12, 
     model = 'L2')
```

```python
pelt(sequence = y, 
     pen = 5.5, 
     model = 'srbf', 
     T = 1, 
     sigma = 1,
     laplace_option = 'dirichlet',
     laplace_hyperparameter = 1,
     kernel_type = 'linear',
     centered = False,
     nonlinear_hyperparam = 1)
```

```python
binseg( sequence = y, 
        n_changepoints = 2, 
        model = 'srbf', 
        T = 1,
        sigma = 1,
        laplace_option = 'dirichlet',
        laplace_hyperparameter = 1,
        kernel_type = 'linear',
        centered = False,
        nonlinear_hyperparam = 1)
```
