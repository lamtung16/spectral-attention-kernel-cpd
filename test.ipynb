```python
import numpy as np
from kernelcpd import spectral_cpd
np.random.seed(123)
```

```python
n_samples = 100
signal1 = np.column_stack((np.random.normal(1.0, 1, n_samples), np.random.normal(1.0, 1, n_samples)))
signal2 = np.column_stack((np.random.normal(5.0, 1, n_samples), np.random.normal(5.0, 1, n_samples)))
y = np.vstack((signal1, signal2))
```

```python
print(spectral_cpd(y=y, T=10))
```
